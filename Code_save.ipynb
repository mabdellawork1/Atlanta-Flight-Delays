```python
import pandas as pd

df = pd.read_csv("US_flights_2023.csv")

print(df.shape)
print(df.columns.tolist())
```

```python
atl = df[(df["Dep_Airport"] == "ATL") | (df["Arr_Airport"] == "ATL")].copy()

print("All US flights:", len(df))
print("ATL flights:",len(atl))
atl.head()
```

```python
atl.info()
```

```python
atl.to_csv("atl_flights_2023.csv", index = False)
```

```python
atl["FlightDate"] = pd.to_datetime(atl["FlightDate"])
atl["Month"] = atl["FlightDate"].dt.month
atl["Month_Name"] = atl["FlightDate"].dt.month_name()
atl = atl.rename(columns={"Aicraft_age": "Aircraft_Age"})
atl = atl.reset_index(drop=True)
atl[["FlightDate", "Month", "Month_Name", "Aircraft_Age"]].head()
```

```python
atl[["Dep_Delay", "Arr_Delay", "Delay_Carrier", "Delay_Weather",
"Delay_NAS", "Delay_Security", "Delay_LastAircraft", "Aircraft_Age"]].describe()
```

```python
atl["Arr_Delay_Tag"] = (atl["Arr_Delay"] >= 15).astype(int)

print("% departures delayed 15+ min:", round(atl["Dep_Delay_Tag"].mean() * 100, 1))
print("% arrivals delayed 15+ min:", round(atl["Arr_Delay_Tag"].mean() * 100, 1))
print("Departures delayed 3+ hours:", (atl["Dep_Delay"] >= 180).sum())
```

```python
atl["Dep_Delay_15"] = (atl["Dep_Delay"] >= 15).astype(int)

print("% departures delayed 15+ min:", round(atl["Dep_Delay_15"].mean() * 100, 1))
print("Smallest delay tagged by the dataset:", atl.loc[atl["Dep_Delay_Tag"] == 1, "Dep_Delay"].min())
```

```python
atl = atl.drop(columns=["Dep_Delay_Tag"])

atl.to_csv("atl_flights_2023_clean.csv", index=False)

print(atl.shape)
```
